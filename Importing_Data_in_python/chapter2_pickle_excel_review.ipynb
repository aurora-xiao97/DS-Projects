## 1. Pickled files

- Pickle is a file format **native to Python**.
- Motivation: many Python objects (dictionaries, custom classes, trained models, and so on) don't have an obvious way to be stored in a flat file.
- **Serialize** means converting an object into a byte stream so it can be saved to disk and later reconstructed exactly as it was.
- Pickled files are usually saved with a `.pkl` extension and must be opened in **binary** mode: `'rb'` to read, `'wb'` to write.
- Use `pickle.load()` inside a `with` block to read the object back into Python.

```python
import pickle

with open('pickled_fruit.pkl', 'rb') as file:
    data = pickle.load(file)

print(data)
print(type(data))

# Expected output:
# {'peaches': 13, 'apples': 4, 'oranges': 11}
# <class 'dict'>
```

## 2. Excel spreadsheets

- Excel files (`.xlsx`) are a common "file from other software" you'll need to import, and unlike a `.csv`, a single workbook can contain **multiple sheets**.
- pandas represents a whole workbook with `pd.ExcelFile(filename)`.
- `.sheet_names` returns a list of every sheet in the workbook.
- `.parse()` turns one sheet into a DataFrame. You can pass either the **sheet name** as a string, or the **sheet index** as an integer (`0` = first sheet).
- You can customize the import the same way you would `pd.read_csv()`:
  - `skiprows` &mdash; skip a number of rows at the top of the sheet
  - `usecols` &mdash; import only specific columns (by index)
  - `names` &mdash; supply your own column names instead of using the sheet's header row

```python
import pandas as pd

file = 'urbanpop.xlsx'
data = pd.ExcelFile(file)
print(data.sheet_names)
# ['1960-1966', '1967-1974', '1975-2011']

df1 = data.parse('1960-1966')   # by sheet name
df2 = data.parse(0)             # by sheet index

# Customized import: skip the header row, keep only the first two
# columns, and give them new names
df3 = data.parse(0, skiprows=1, usecols=[0, 1], names=['country', 'population'])
print(df3.head())
```

Pickle is useful for objects that don't map naturally onto a flat file's rows and columns, such as dictionaries, trained machine learning models, or custom class instances.

**A3.** Pickle files are binary, not text, so opening one in text mode (`'r'`) would either raise a `UnicodeDecodeError` or silently produce garbled, unusable data, since Python would try to decode raw bytes as text characters.
