## Chapter 1 — Dates and Calendars

### 1. Creating a `date` object
Import `date` from the built-in `datetime` module.  
Arguments: `date(year, month, day)`

```python
from datetime import date

# Create a single date
d = date(2017, 6, 21)
print(d)          # 2017-06-21
print(type(d))    # <class 'datetime.date'>
```

### 2. Date attributes
Access individual components with `.year`, `.month`, and `.day`.

```python
d = date(2016, 10, 7)

print(d.year)    # 2016
print(d.month)   # 10
print(d.day)     # 7
```

### 3. Finding the weekday
`weekday()` returns an integer: **0 = Monday → 6 = Sunday**

```python
d = date(2016, 10, 7)
print(d.weekday())   # 4 = Friday

# Mapping for reference
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
print(days[d.weekday()])   # Friday
```

### 4. Math with dates — `timedelta`
Subtracting two `date` objects gives a `timedelta`.  
You can also add a `timedelta` to shift a date forward or backward.

```python
from datetime import date, timedelta

d1 = date(2017, 11, 5)
d2 = date(2017, 12, 4)

# Subtract two dates → timedelta
delta = d2 - d1
print(delta)        # 29 days, 0:00:00
print(delta.days)   # 29

# Add a timedelta to a date
td = timedelta(days=29)
print(d1 + td)      # 2017-12-04

# Find the earliest date in a list
dates = [date(2017, 12, 4), date(2017, 11, 5)]
print(min(dates))   # 2017-11-05
```

### 5. Turning dates into strings

**Two methods:**
- `.isoformat()` → always returns `YYYY-MM-DD` (no arguments needed)
- `.strftime(format)` → returns any format you define using format codes

| Code | Meaning | Example |
|------|---------|--------|
| `%Y` | 4-digit year | 2017 |
| `%y` | 2-digit year | 17 |
| `%m` | Month, zero-padded | 06 |
| `%B` | Full month name | June |
| `%b` | Short month name | Jun |
| `%d` | Day, zero-padded | 05 |
| `%A` | Full weekday name | Tuesday |
| `%a` | Short weekday name | Tue |

```python
from datetime import date

d = date(2017, 6, 5)

# isoformat — always YYYY-MM-DD
print(d.isoformat())               # 2017-06-05

# strftime — your custom format
print(d.strftime("%Y/%m/%d"))      # 2017/06/05
print(d.strftime("%d-%b-%Y"))      # 05-Jun-2017
print(d.strftime("%B %d, %Y"))     # June 05, 2017
print(d.strftime("Year is %Y"))    # Year is 2017
print(d.strftime("%A"))            # Monday

# IMPORTANT: strftime returns a NEW string — d itself stays a date object
result = d.strftime("%Y-%m-%d")
print(type(d))       # <class 'datetime.date'>   ← unchanged
print(type(result))  # <class 'str'>              ← new string
```

## Chapter 2 — Combining Dates and Times

### 6. Creating a `datetime` object
A `datetime` holds both date AND time.  
Arguments: `datetime(year, month, day, hour, minute, second, microsecond)`  
Only year, month, day are required — time parts default to 0.

```python
from datetime import datetime

# Positional arguments
dt = datetime(2017, 10, 1, 15, 23, 25, 500000)
print(dt)   # 2017-10-01 15:23:25.500000

# Keyword arguments (clearer to read)
dt = datetime(year=2017, month=10, day=1,
              hour=15, minute=23, second=25,
              microsecond=500000)
print(dt)   # 2017-10-01 15:23:25.500000

# Access any component
print(dt.year, dt.month, dt.day)          # 2017 10 1
print(dt.hour, dt.minute, dt.second)      # 15 23 25
```

### 7. Replacing parts of a datetime
`replace()` creates a new datetime with only the fields you specify changed.  
The original object is never modified.

```python
from datetime import datetime

dt = datetime(2017, 10, 1, 15, 23, 25, 500000)
print(dt)       # 2017-10-01 15:23:25.500000

# Round down to the nearest hour
dt_hr = dt.replace(minute=0, second=0, microsecond=0)
print(dt_hr)    # 2017-10-01 15:00:00

# Original is unchanged
print(dt)       # 2017-10-01 15:23:25.500000
```

### 8. Formatting datetimes with `strftime`
Same as for `date`, but you can now also use time format codes.

| Code | Meaning | Example |
|------|---------|--------|
| `%H` | Hour, 24-hour | 15 |
| `%I` | Hour, 12-hour | 03 |
| `%p` | AM / PM | PM |
| `%M` | Minute | 23 |
| `%S` | Second | 25 |
| `%f` | Microseconds | 500000 |

```python
from datetime import datetime

dt = datetime(2017, 12, 30, 15, 19, 13)

# Date only
print(dt.strftime("%Y-%m-%d"))             # 2017-12-30

# Date and time
print(dt.strftime("%Y-%m-%d %H:%M:%S"))    # 2017-12-30 15:19:13

# Custom layout
print(dt.strftime("%H:%M:%S on %Y/%m/%d")) # 15:19:13 on 2017/12/30

# 12-hour clock
print(dt.strftime("%I:%M %p"))             # 03:19 PM

# ISO 8601 format (T separator between date and time)
print(dt.isoformat())                       # 2017-12-30T15:19:13
```

### 9. Parsing strings with `strptime`
`strptime` = **string parse time** → converts a string INTO a `datetime` object.  
You must provide the format string that matches the input exactly.

| Method | Direction | Use when... |
|--------|-----------|-------------|
| `strftime` | datetime → string | Displaying / printing a date |
| `strptime` | string → datetime | Reading dates from CSV / user input |

```python
from datetime import datetime

# Parse a US-format date string
dt = datetime.strptime("12/30/2017 15:19:13", "%m/%d/%Y %H:%M:%S")
print(type(dt))   # <class 'datetime.datetime'>
print(dt)         # 2017-12-30 15:19:13
print(dt.year)    # 2017  ← now you can use all datetime attributes!

# Parse an ISO-format string
dt2 = datetime.strptime("2017-06-05", "%Y-%m-%d")
print(dt2)        # 2017-06-05 00:00:00

# Common real-world use: dates loaded from a CSV come in as strings
date_from_csv = "2017-06-05"
dt3 = datetime.strptime(date_from_csv, "%Y-%m-%d")
print(dt3 + __import__('datetime').timedelta(days=1))  # 2017-06-06 00:00:00
```

```python
from datetime import timedelta
```

```python
print(dt3 + timedelta(days=1))  # 2017-06-06 00:00:00
```

```python
# WARNING: format string must match the input EXACTLY
# This will raise a ValueError because the time part is not covered
try:
    dt = datetime.strptime("2017-12-30 15:19:13", "%Y-%m-%d")
except ValueError as e:
    print(f"Error: {e}")   # unconverted data remains: 15:19:13
```

### 10. Converting from a Unix timestamp
`datetime.fromtimestamp()` converts a Unix timestamp (seconds since Jan 1 1970) into a `datetime`.

```python
from datetime import datetime

ts = 1514665153.0
dt = datetime.fromtimestamp(ts)
print(dt)   # 2017-12-30 15:19:13
```

### 11. Working with durations — `timedelta` with datetimes
Subtracting two `datetime` objects gives a `timedelta`.  
Use `.total_seconds()` to get the full duration in seconds.

```python
from datetime import datetime, timedelta

start = datetime(2017, 10, 8, 23, 46, 47)
end   = datetime(2017, 10, 9,  0, 10, 57)

# Subtract → timedelta
duration = end - start
print(duration)                   # 0:24:10
print(duration.total_seconds())   # 1450.0

# Create timedeltas manually
delta1 = timedelta(seconds=1)             # one second
delta2 = timedelta(days=1, seconds=1)     # one day and one second
delta3 = timedelta(weeks=-1)              # negative = go backwards

print(start + delta1)    # 2017-10-08 23:46:48
print(start + delta2)    # 2017-10-09 23:46:48
print(start + delta3)    # 2017-10-01 23:46:47  (one week earlier)
print(start - timedelta(weeks=1))  # same result as above
```
