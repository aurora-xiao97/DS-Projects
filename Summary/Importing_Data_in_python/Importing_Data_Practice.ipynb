## 4. Importing flat files with NumPy

NumPy arrays are the standard way to store **purely numerical** data, and are required by
packages like scikit-learn.

```python
import numpy as np
data = np.loadtxt('MNIST.txt', delimiter=',')
```

Useful arguments:

| Argument | Purpose |
|---|---|
| `delimiter` | character that separates values |
| `skiprows` | number of rows to skip (e.g. header) |
| `usecols` | list of column **indices** to keep |
| `dtype=str` | force every value to load as a string |

⚠️ **Limitation:** a single NumPy array can only hold **one dtype**. If a file mixes numbers and
strings (like `titanic.csv` with `Name`, `Sex`, `Fare`), `np.loadtxt()` either fails or forces
everything to one type (e.g. all strings) — this is where pandas comes in.


## 5. Importing flat files with pandas

A pandas **DataFrame** is a 2D labeled data structure where **each column can have its own dtype**
— exactly what most real-world flat files need.

```python
import pandas as pd
data = pd.read_csv('winequality-red.csv')
data.head()                 # preview first 5 rows
data_array = data.to_numpy()  # convert to a NumPy array if needed
```

pandas is the standard tool for exploratory data analysis, data wrangling/preprocessing,
building models, and visualization.

### Solution 4

````python
# a) without dtype, this raises a ValueError because np.loadtxt can't convert
#    strings like "Braund" or "male" to floats (its default dtype).
try:
    np.loadtxt('mini_titanic.csv', delimiter=',', skiprows=1)
except ValueError as e:
    print("ValueError:", e)

# b) dtype=str works, but now EVERY column (including Age and Fare) is a string,
#    so you lose the ability to do numeric operations (e.g. mean()) without converting back.
data_str = np.loadtxt('mini_titanic.csv', delimiter=',', skiprows=1, dtype=str)
print(data_str)

# c) pandas infers each column's type independently
import pandas as pd
df_check = pd.read_csv('mini_titanic.csv')
print(df_check.dtypes)
# Name/Sex -> object (string), Age -> int64, Fare -> float64
# This is exactly the "columns of potentially different types" pandas is built for.
````

### Quiz Answers

1. `np.loadtxt()` forces the **entire array** to one dtype (it will error on mixed types unless
   you force `dtype=str`, which then loses numeric typing). `pd.read_csv()` automatically infers
   a **separate dtype per column**, so numbers stay numeric and text stays text in the same DataFrame.
2. The context manager automatically closes the file for you — even if an error happens inside
   the `with` block — so you never accidentally leave a file handle open.
3. `skiprows=n` skips the first *n* rows of the file (e.g. a header row). `usecols=[...]` keeps
   only the columns at the given index positions, discarding the rest.
4. **True.** Each column in a DataFrame can hold its own dtype (e.g. `Name` as strings, `Age` as
   integers, `Fare` as floats) — this mirrors real data like `titanic.csv`, which is exactly why
   pandas (not NumPy) is the standard tool for files with mixed data types.
5. Flat files, **files from other software** (e.g. Excel, SAS, Stata), and **relational databases**.
