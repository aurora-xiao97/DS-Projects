```python
from scipy import stats
import numpy as np

def quest_retention_analysis(prior_alpha, prior_beta, trials, successes):
    """
    Run a Bayesian update on day-1 retention for the Daily Quest feature.

    Returns a dict with:
      - post_alpha   : updated alpha
      - post_beta    : updated beta
      - prior_mean   : prior belief about retention
      - post_mean    : posterior mean retention rate
      - post_variance: posterior variance
      - ci_95        : (lower, upper) 95% credible interval
      - lift         : posterior mean - prior mean
      - is_promising : True if entire 95% CI is above the prior mean
    """
    # YOUR CODE HERE
    assert prior_alpha > 0 and prior_beta > 0
    assert trials >= 0
    assert 0 <= successes <= trials

    post_alpha = prior_alpha + successes
    post_beta = prior_beta + (trials - successes)
    prior_mean = prior_alpha/(prior_alpha + prior_beta)
    post_mean  = post_alpha/(post_alpha + post_beta)
    s = post_alpha + post_beta
    post_variance = post_alpha * post_beta / (s**2 *(s+1))
    ci_95_lower = stats.beta.ppf(0.025, post_alpha, post_beta)
    ci_95_upper = stats.beta.ppf(0.975, post_alpha, post_beta)
    ci_95 = (ci_95_lower, ci_95_upper)
    lift = post_mean - prior_mean
    is_promising = bool(ci_95_lower > prior_mean)
  
        # ── Output validation ─────────────────────────────────
    assert 0 < prior_mean < 1,            "prior_mean must be in (0,1)"
    assert 0 < post_mean < 1,             "post_mean must be in (0,1)"
    assert post_variance > 0,             "variance must be positive"
    assert 0 < ci_95_lower < ci_95_upper < 1, "CI must be ordered and in (0,1)"
    assert isinstance(is_promising, bool), "is_promising must be bool"


    return {'post_alpha': post_alpha, 
            'post_beta': post_beta,
            'prior_mean': prior_mean,
            'post_mean': post_mean,
            'post_variance': post_variance,
            'ci_95': ci_95,
            'lift': lift,
            'is_promising': is_promising
            }


```

```python
# ── Run it ──────────────────────────────────────────────
result = quest_retention_analysis(
    prior_alpha = 3,
    prior_beta  = 7,
    trials      = 500,
    successes   = 178
)

for k, v in result.items():
    print(f"{k:>15}: {v}")
```
