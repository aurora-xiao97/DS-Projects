## Working with Dates and Times in Python

```python
## importing the date function within the datetime package

from datetime import date
```

```python
### to create a today object

today = date(2026, 6,20)
```

```python
today
```

```python
## if i want to know the week of day of the today
## python index starting from 0, with 0 being Monday, then 5 in this case is Saturday
print(today.weekday())
```

```python
### using attributes like .year, .month, .day to get relevant date information
print(today.year)
print(today.month)
print(today.day)
```

```python
# Import date
from datetime import date

# Create a date object for May 9th, 2007
start = date(2007, 5, 9)

# Create a date object for December 13th, 2007
end = date(2007, 12, 13)

# Subtract the two dates and print the number of days
print((end - start).days)
```

```python
# Import date
from datetime import date

# Create a date object
andrew = date(1992, 8, 26)

# Print the date in the format 'YYYY-MM'
print(andrew.strftime("%Y-%m"))


# Create a date object
andrew = date(1992, 8, 26)

# Print the date in the format 'MONTH (YYYY)'
print(andrew.strftime("%B (%Y)"))

# Create a date object
andrew = date(1992, 8, 26)

# Print the date in the format 'YYYY-DDD'
print(andrew.strftime("%Y-%j"))
```
