### Probability Foundations & Bayesian Reasoning

```python
from scipy import stats

# Likelihood: how probable is seeing 30 clicks if true CTR = 10%?
likelihood = stats.binom.pmf(k=30, n=200, p=0.10)
print(f"P(D | H=0.10) = {likelihood:.6f}")   # very small number ~0.0419

# Compare: how likely if true CTR were 15% instead?
likelihood_15 = stats.binom.pmf(k=30, n=200, p=0.15)
print(f"P(D | H=0.15) = {likelihood_15:.6f}")  # larger — data fits 15% better
```
