```python
import numpy as np
```

```python
vectorAsList = [1, 2, 3, 4, 5]
vectorAsNumpyArray = np.array([1, 2, 3, 4, 5])
rowVectorAsNumpyArray = np.array([[1, 2, 3, 4, 5], [2, 3, 4, 5, 6]])
columnVectorAsNumpyArray = np.array([[1], [2], [3], [4], [5]])
```

```python
print("vectorAsList:", vectorAsList)
print("vectorAsNumpyArray:", vectorAsNumpyArray) 
print("rowVectorAsNumpyArray:", rowVectorAsNumpyArray)
print("columnVectorAsNumpyArray:", columnVectorAsNumpyArray)
```

```python
### vector addition:

vector1 = np.array([1, 2, 3])
vector2 = np.array([4, 5, 6])
vector_sum = vector1 + vector2
```

```python
print(vector_sum)  # Output: [5 7 9]
```

```python
### vector subtraction:
vector_diff = vector2 - vector1 
print(vector_diff)  # Output: [-3 -3 -3]
```

### The Dot Product

```python
np.dot(vector1, vector2)  # Output: 32
```
